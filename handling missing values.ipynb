```python
import pandas as pd
```

```python
data = {
    "Name": ["  Ali", "Sara ", "JOHN", "Ali", "Riya", None],
    "Age": [20, 21, "22", 20, "unknown", 23],
    "City": ["Rajkot", " rajkot", "Mumbai", "Rajkot", None, "Delhi"],
    "Gender": ["Male", "female", "MALE", "Male", "F", "Female"],
    "Salary": [25000, 30000, None, 25000, 40000, 35000],
    "Empty_Column": [None, None, None, None, None, None]
}

df = pd.DataFrame(data)

df.head()
```

```python
#inspect the data
df.shape
```

```python
df.info()
```

**2.Handling Missing Values**

```python
df.isnull()
```

```python
df.isnull().sum()
```

**3. Fill missing values**


```python
df["Name"] = df["Name"].fillna("Unknown")

df
```

```python
df["City"] = df["City"].fillna("Unknown")

df
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df
```

**5. Fix mixed data types**

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

df
```
